```python
# Keep all columns as pandas string dtype so IDs/zip/phone values are not coerced to float.
data_and_voter_data = pd.concat([data, voter_data], ignore_index=True).astype("string")
df = data_and_voter_data.groupby("cluster_id", dropna=False).first(skipna=True).reset_index()
print(len(df))
```

```python
# filter out schools
df = df[~df["company_name"].str.lower().str.contains("school|elementary|university|college")]
print(len(df))
```

```python
# Create metrics for each dataset
datasets = {'data': data, 'voter_data': voter_data, 'df': df}

# Get all unique columns across all datasets
all_columns = set()
for name, dataset in datasets.items():
    all_columns.update(dataset.columns)

# Calculate fill rate for each column in each dataset
metrics = {}
for name, dataset in datasets.items():
    row_metrics = {}
    for col in all_columns:
        if col in dataset.columns:
            non_null_count = dataset[col].notna().sum()
            row_metrics[col] = non_null_count / 1044781
        else:
            row_metrics[col] = 0.0
    metrics[name] = row_metrics

# Create metrics dataframe
metrics_df = pd.DataFrame(metrics).T
metrics_df.index.name = 'dataset'
metrics_df = metrics_df.reset_index()
metrics_df
```

```python
metrics_df_melted = metrics_df.melt(id_vars='dataset', var_name='column', value_name='fill_rate')
metrics_df_pivoted = metrics_df_melted.pivot(index='column', columns='dataset', values='fill_rate').reset_index()
metrics_df_pivoted = metrics_df_pivoted[["column", "data", "voter_data", "df"]]
for col in metrics_df_pivoted.columns.tolist():
    if col != "column":
        metrics_df_pivoted[col] = metrics_df_pivoted[col].round(2)
metrics_df_pivoted
```

```python
metrics_df_pivoted = metrics_df_pivoted[metrics_df_pivoted["df"] <= metrics_df_pivoted["data"]]
metrics_df_pivoted
```

```python
# Safety cleanup for export: convert values like 123.0 -> 123 while keeping true decimals intact.
def _strip_integer_dot_zero(value):
    if pd.isna(value):
        return pd.NA
    s = str(value)
    if s.endswith(".0") and s[:-2].lstrip("-").isdigit():
        return s[:-2]
    return s

export_df = df.apply(lambda col: col.map(_strip_integer_dot_zero)).astype("string")
export_df.to_csv("csv_outputs/data_appended_voter_info_2026-04-13.csv", index=False)
metrics_df_pivoted.to_csv("csv_outputs/metrics_post_appended_voter_info_2026-04-13.csv", index=False)
```
