```python
import pandas as pd

data = pd.read_csv("csv_outputs/deduped_data_domestic_2026-01-14.csv", dtype="string")
# data = pd.read_csv("csv_outputs/data_appended_voter_info_2026-02-07.csv", dtype="string")
```

```python
data.columns.tolist()
```

```python
home_address_cols = ["home_address","home_city","home_state","home_zipcode","home_zipfour"]
home_address_mask = data[home_address_cols].isna().all(axis=1)

party_mask = data["party_description"].isna()

voting_cols = ["primary_n_of_4","general_2024","primary_2024","general_2022","primary_2022","general_2020","primary_2020","voting_performance_minor_election","voting_performance_even_year_general","voting_performance_even_year_primary","voting_performance_even_year_general_and_primary","us_congressional_district","state_senate_district","state_house_district","state_legislative_district","county_commissioner_district","county_supervisorial_district","precinct","state_voter_id"]
voting_mask = data[voting_cols].isna().all(axis=1)

final_filter = home_address_mask & party_mask & voting_mask
reversed_filter = ~final_filter
```

```python
data_no_voting_filter = data[final_filter]
data_yes_voting_filter = data[reversed_filter]

cols_to_keep = ["cluster_id", "first_name", "last_name", "mobile_phone", "company_city", "company_state", "company_zipcode"]
data_no_voting_filter = data_no_voting_filter[cols_to_keep]
data_yes_voting_filter = data_yes_voting_filter[cols_to_keep]
```

```python
print(len(data_yes_voting_filter)/len(data))
```

```python
from datetime import datetime
datetime_now = datetime.now().strftime("%Y-%m-%d")
data_no_voting_filter.to_csv(f"csv_outputs/data_no_voting_filter_{datetime_now}.csv", index=False)
data_yes_voting_filter.to_csv(f"csv_outputs/data_yes_voting_filter_{datetime_now}.csv", index=False)
```
