```python
# merged state data into one masterfile
import pandas as pd
import os

raw_data_folder = 'raw_data_states'
excel_files = []
for root, dirs, files in os.walk(raw_data_folder):
    for file in files:
        if file.endswith('.xlsx') or file.endswith('.xls'):
            excel_files.append(os.path.join(root, file))

masterfile_columns = pd.read_csv('columns_to_keep/state_file_columns.csv')['column_names'].tolist()

df = pd.DataFrame()
for file in excel_files:
    print(file)
    temp_df = pd.read_excel(file, dtype="string")
    temp_df = temp_df[masterfile_columns]
    temp_df["SOURCEFILE"] =  '[STATE] ' + file
    df = pd.concat([df, temp_df], ignore_index=True)

df = df.drop_duplicates(subset=[x for x in df.columns.tolist() if x != "SOURCEFILE"])
```
